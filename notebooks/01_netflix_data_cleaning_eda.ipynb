## Data Quality Assessment Summary

This section consolidates the data-quality checks performed on the Netflix dataset. The assessment examines missing values, duplicate records, data types, formatting consistency, and potential inconsistencies across columns. The findings will guide the cleaning decisions before exploratory data analysis (EDA).

```python
quality_summary = pd.DataFrame({
    "Column": df.columns,
    "Missing Values": df.isna().sum().values,
    "Missing Percentage": (df.isna().mean().values * 100).round(2)
})

quality_summary
```

```python
# Check for duplicate rows
df.duplicated().sum()
```

**Initial Data Quality Findings**

- **Dataset size:** 200 records and 12 columns.
- **Missing values:** The `director` column has 63 missing values (31.5%), `cast` has 20 (10%), and `country` has 64 (32%).
- **Duplicate records:** No duplicate rows were identified.
- **Other columns:** The remaining nine columns contain no missing values.

Missing information will be assessed before cleaning. Records will not be deleted, and missing values will not be guessed or filled without a justified reason.
